```python
def fixed(text, size=300):
    return [text[i : i + size] for i in range(0 , len(text), size)]


def fixed_overlap(text, size=300, overlap = 80):
    step = size-overlap
    return [text[i : i+size] for i in range(0, len(text), step )]

def recursive(text, size=300):
    paras = [ p.strip() for p in text.split("\n\n") if p.strip()]
    chunks = [ ]
    for p in paras:
        if len(p) <= size:
            chunks.append(p)
        else:
            sents, cur = re.split(r'(?<=[.!?])\s+',p), ""
            for s in sents:
                if len(cur) + len(s) <= size:
                    cur += " " + s
                else:
                    chunks.append(cur.strip())
                    cur = s
            if cur.strip():
                chunks.append(cur.strip())
    return chunks

def semantic(text, threshold=0.5):
    sents = [s.strip() for s in re.split(r'(?<=[.!?])\s+', text.replace("\n"," ")) if s.strip()]
    embs = bi.encode(sents, normalize_embeddings=True)
    chunks, cur = [], [sents[0]]
    for i in range(1, len(sents)):
        sim = float(embs[i]@embs[i-1])
        if sim < threshold:
            chunks.append(" ".join(cur))
            cur = [sents[i]]
        else:
            cur.append(sents[i])
    if cur: chunks.append(" ".join(cur))
    return chunks


strategies = {
    "fixed": fixed(document),
    "fixed_overlap" : fixed_overlap(document),
    "recursive" : recursive(document),
    "semantic" : semantic(document),
}

for name, chunks in strategies.items():
    print(f"{name:14} {len(chunks)} chunks | avg len {int(np.mean([len(c) for c in chunks]))}")
```

```python
indexes = {name : bi.encode(chunks, normalize_embeddings=True) for name, chunks in strategies.items()}

def retrieve(query, name, k=3):
    emb, chunks = indexes[name], strategies[name]
    q = bi.encode([query], normalize_embeddings=True)[0]
    top = np.argsort(-(emb @ q))[:k]
    return [chunks[i] for i in top]
```

```python
queries = [
    "how long is onboarding", "what are the royalty fees", "when are royalties invoiced",
    "what is the late payment charge", "how long is the equipment warranty",
    "how often is maintenance required", "minimum store hours per week",
    "how do I request a temporary closure", "how long do refunds take",
    "what happens to returns without a receipt", "who approves large refunds",
    "what percent is the marketing contribution", "how do I update store hours",
    "where must replacement parts come from", "how do I dispute a royalty charge",
]

for q in queries[:5]:   # eyeball a few first
    print(f"\n\n Q: {q}")
    for name in strategies:
        top1 = retrieve(q, name, k=1)[0]
        print(f"\n {name:14} → {top1} -> length : {len(top1)}")
```

```python
def top_terms(chunk):
    return set(re.findall(r'\w+', chunk.lower()))

def overlap_vs_reference(query, ref="recursive", k=3):
    ref_terms = set().union(*[top_terms(c) for c in retrieve(query, ref, k)])
    row = {}
    for name in strategies:
        got = set().union(*[top_terms(c) for c in retrieve(query, name, k)])
        row[name] = len(got & ref_terms) / len(got | ref_terms)
    return row

agg = {name : [] for name in strategies}
for q in queries:
    for name, val in overlap_vs_reference(q).items():
        agg[name].append(val)

print(" avg content-overlap vs 'recursive' baseline(1.0 = identical retrieval) : ")
for name in strategies:
    print(f"{name:14} {statistics.mean(agg[name]):.2f}")
```

```python
bi2 = SentenceTransformer("BAAI/bge-small-en-v1.5")

ref_chunks = strategies["recursive"]
idx_a = bi.encode(ref_chunks, normalize_embeddings=True)
idx_b = bi2.encode(ref_chunks, normalize_embeddings=True)

def retrieve_terms(query, model, index, k=3):
    q = model.encode([query], normalize_embeddings=True)[0]
    top =np.argsort(-(index @ q))[:k]
    return set().union(*[top_terms(ref_chunks[i]) for i in top])

model_overlaps = []
for q in queries:
    a = retrieve_terms(q,bi,idx_a)
    b=  retrieve_terms(q, bi2, idx_b)
    model_overlaps.append(len(a & b) / len( a | b))

model_swap = statistics.mean(model_overlaps)
chunk_swap = statistics.mean(agg["fixed"])

print(f"overlap when swapping EMBEDDING MODEL (chunking fixed): {model_swap:.2f}")
print(f"overlap when swapping CHUNKING (model fixed):           {chunk_swap:.2f}")
print(f"\n-> lower overlap = bigger disruption to retrieval")
print(f"-> chunking moved retrieval {'MORE' if chunk_swap < model_swap else 'LESS'} than the embedding model")
```

```python
def retrieve_terms_bge(query, model, index, k=3):
    q = model.encode(["Represent this sentence for searching relevant passages: " + query],
                     normalize_embeddings=True)[0]
    top = np.argsort(-(index @ q))[:k]
    return set().union(*[top_terms(ref_chunks[i]) for i in top])

model_overlaps = []
for q in queries:
    a = retrieve_terms(q, bi, idx_a)          # MiniLM, no prefix
    b = retrieve_terms_bge(q, bi2, idx_b)     # BGE, with prefix
    model_overlaps.append(len(a & b) / len(a | b))
print(f"model swap (BGE prefixed): {statistics.mean(model_overlaps):.2f}")
```

### Lab 9 — Chunking Effects (Observation Note)

#### What this lab was about
Measuring how much the CHUNKING strategy determines retrieval, vs how much the EMBEDDING MODEL does. Chunked one ~4k-word structured doc four ways, embedded each, ran 15 queries, and measured content-overlap (Jaccard on retrieved words) to quantify how differently each setup retrieves.

#### The four strategies (cheap/structure-blind → expensive/meaning-aware)
1. Fixed-size, no overlap — cut every N chars; guillotines sentences, chunks start mid-topic.
2. Fixed-size, with overlap — repeats ~80 chars across boundaries; saves boundary-straddling facts but imports neighbor noise.
3. Recursive/structural — split on paragraph→sentence boundaries; respects structure. Production standard (LangChain's RecursiveCharacterTextSplitter).
4. Semantic — embed sentences, cut where consecutive-sentence similarity drops; topically coherent chunks. Smartest, but can over-fragment (some chunks were 76 chars = precise retrieval but thin context).

#### Measured: how much chunking moves retrieval (overlap vs recursive baseline)
- fixed          0.45   (retrieves ~45% same content as recursive → big divergence)
- fixed_overlap  0.42
- recursive      1.00   (baseline)
- semantic       0.68   (closest to recursive, but still ~1/3 different)
Naive fixed-size is the biggest departure; semantic clusters nearest to recursive.

#### The headline: chunking vs embedding model
- Swap embedding model (MiniLM→BGE), chunking fixed:        0.76 overlap (0.81 with BGE query prefix)
- Swap chunking (fixed→recursive), model fixed:             0.45 overlap
- Fair version (semantic→recursive, good-vs-good):          0.68 overlap
Lower overlap = bigger disruption. Chunking (0.45, or 0.68 fair) disrupted retrieval MORE than swapping the embedding model (0.76–0.81). **Chunking is the higher-leverage decision.**

#### Key correction / methodology lessons
- BGE needs a query prefix ("Represent this sentence for searching relevant passages: ") — adding it raised overlap 0.76→0.81, confirming it calibrates BGE's retrieval. Forgetting it silently degrades results.
- A silently-swallowed kwarg typo (normalize_embeddings misspelled) and a copy-paste bug (indexing both sides with the same model) produced a fake "model matters more" result first. Lesson: sentence-transformers ignores unknown kwargs silently — verify your two conditions are actually different before trusting a comparison.
- "I don't know"-style low overlap can be an artifact; always confirm the pipeline is measuring what you think.

#### Magnitude is document-dependent (nuanced interview answer)
On this clean, well-structured doc, even fixed chunks land in roughly the right section, so chunking has less room to wreck things — yet it STILL out-disrupted the model swap. On messy/poorly-structured corpora the gap would be larger. So: chunking generally dominates, but I'd profile the actual corpus rather than assume.

Portal Assistant uses one-task-per-file, <600-token chunks — effectively a semantic/structural strategy (each file = one coherent unit). This lab is evidence that was a higher-leverage choice than which embedder I picked. Good instinct, now measured.

#### One-line summary
Chunking strategy moves retrieval more than the embedding model does (0.45 vs 0.81 overlap on my doc) — engineers overinvest in "which embedding model" and underinvest in chunking, which is backwards.
